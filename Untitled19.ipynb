## 1. Introduction to Pandas

Pandas is a powerful and popular open-source Python library used for data manipulation and analysis. It provides data structures like DataFrames and Series, which are ideal for handling structured data. It's built on top of NumPy, offering high-performance operations.

### Why use Pandas?

*   **Easy to use**: Intuitive API for data operations.
*   **Fast**: Highly optimized for performance.
*   **Flexible**: Handles various data types and formats.
*   **Powerful**: Offers extensive functionalities for cleaning, transforming, and analyzing data.

Let's start by importing the pandas library.

```python
import pandas as pd
print("Pandas imported successfully!")
```

## 2. Pandas Series

A Pandas Series is a one-dimensional array-like object capable of holding any data type (integers, strings, floats, Python objects, etc.). It has an associated array of data labels, called its index.

Think of a Series as a single column of data in a spreadsheet or a SQL table.

### Creating a Series

```python
# Create a Series from a list
data = [10, 20, 30, 40, 50]
s = pd.Series(data)
print("Series from list:")
print(s)

print("\n")

# Create a Series with a custom index
data_with_index = [100, 200, 300]
index_labels = ['a', 'b', 'c']
s_indexed = pd.Series(data_with_index, index=index_labels)
print("Series with custom index:")
print(s_indexed)

print("\n")

# Create a Series from a dictionary
dict_data = {'apple': 10, 'banana': 20, 'cherry': 30}
s_dict = pd.Series(dict_data)
print("Series from dictionary:")
print(s_dict)
```

### Accessing Elements in a Series
You can access elements in a Series using their integer position or their index label.

```python
# Access by position
print(f"Element at position 0 (s): {s[0]}")

# Access by label
print(f"Element with label 'b' (s_indexed): {s_indexed['b']}")

# Access multiple elements
print("\nMultiple elements by position (s):")
print(s[[1, 3]])

print("\nMultiple elements by label (s_indexed):")
print(s_indexed[['a', 'c']])
```

## 3. Pandas DataFrame

A Pandas DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns).

Think of a DataFrame as a spreadsheet, a SQL table, or a dictionary of Series objects.

### Creating a DataFrame
DataFrames can be created from various sources, such as dictionaries, lists of lists, or external files (CSV, Excel, etc.).

```python
# Create a DataFrame from a dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David','rahul'],
    'Age': [24, 27, 22, 32,23],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston','Noida']
}
df = pd.DataFrame(data)
print("DataFrame from dictionary:")
print(df)
print(
    
)
print(data["Age"])
print("\n")

# Create a DataFrame with custom row index
df_indexed = pd.DataFrame(data, index=['a', 'b', 'c', 'd','e'])
print("DataFrame with custom index:")
print(df_indexed)

print("\n")

# Create a DataFrame from a list of lists (providing column names)
data_list = [
    ['Eve', 29, 'Boston'],
    ['Frank', 35, 'Seattle']
]
columns_list = ['Name', 'Age', 'City']
df_list = pd.DataFrame(data_list, columns=columns_list)
print("DataFrame from list of lists:")
print(df_list)
```

### Accessing Rows (Selection with `loc` and `iloc`)
Pandas provides two primary methods for row selection:

*   `.loc[]`: Label-based indexing (uses row labels).
*   `.iloc[]`: Integer-location based indexing (uses integer positions).

```python
# Select row by integer position (iloc)
print("Row at index 0 (iloc[0]):")
print(df.iloc[0])

print("\nRow with custom label 'b' (loc['b'], using df_indexed):")
print(df_indexed.loc['b'])

# Select multiple rows by integer position
print("\nRows at index 1 and 3 (iloc[[1, 3]]):")
print(df.iloc[[1, 3]])

# Select rows and specific columns
print("\nName and Age for rows at index 0 and 2 (iloc):")
print(df.iloc[[0, 2], [0, 1]]) # First row, third row; first col, second col

print("\nName and Age for custom labels 'a' and 'c' (loc):")
print(df_indexed.loc[['a', 'c'], ['Name', 'Age']])
```

## 4. Data Manipulation

Data manipulation involves cleaning, transforming, and preparing data for analysis. This includes filtering, sorting, adding/removing columns, and handling missing values.

### Filtering Data
You can filter DataFrames based on conditions applied to one or more columns.

```python
# Filter rows where Age > 25
young_adults = df[df['Age'] > 25]
print("People older than 25:")
print(young_adults)

print("\n")

# Filter rows with multiple conditions
# People older than 25 AND from New York
condition = (df['Age'] < 25) & (df['City'] == 'Noida')
filtered_data = df[condition]
print("People older than 25 from New York:")
print(filtered_data)
```

### Adding and Modifying Columns
It's common to add new columns derived from existing ones or modify existing columns.

```python
# Add a new column 'IsAdult'
df['IsAdult'] = df['Age'] >= 18
print("DataFrame with 'IsAdult' column:")
print(df)

print("\n")

# Modify an existing column
df['Age'] = df['Age'] + 1 # Increment all ages by 1
print("DataFrame with incremented ages:")
print(df)
```

### Handling Missing Values
Missing values (NaN - Not a Number) are common in real-world datasets. Pandas provides methods to detect and handle them.

```python
import numpy as np

df_missing = pd.DataFrame({
    'A': [1, 2, np.nan, 4],
    'B': [np.nan, 6, 7, 8],
    'C': [9, 10, 11, np.nan]
})
print("DataFrame with missing values:")
print(df_missing)

print("\nCheck for null values (df_missing.isnull()):")
print(df_missing.isnull())

print("\nCount null values per column (df_missing.isnull().sum()):")
print(df_missing.isnull().sum())

# Drop rows with any missing values
df_dropped_rows = df_missing.dropna()
print("\nDataFrame after dropping rows with NaN:")
print(df_dropped_rows)

# Fill missing values with a specific value (e.g., 0)
df_filled_zero = df_missing.fillna(0)
print("\nDataFrame after filling NaN with 0:")
print(df_filled_zero)

# Fill missing values with the mean of the column
df_filled_mean = df_missing.fillna(df_missing.mean(numeric_only=True))
print("\nDataFrame after filling NaN with column mean:")
print(df_filled_mean)
```

## 5. Data Analysis

Data analysis with Pandas often involves grouping data, performing aggregations, and calculating statistics to gain insights.

### Grouping Data (`groupby()`)
The `groupby()` method is used to group rows together based on one or more column values and then apply an aggregation function (like `sum()`, `mean()`, `count()`, etc.) to the groups.

```python
sales_data = pd.DataFrame({
    'Product': ['A', 'B', 'A', 'C', 'B', 'A', 'C'],
    'Region': ['East', 'West', 'East', 'North', 'West', 'East', 'North'],
    'Sales': [100, 150, 120, 90, 180, 110, 95]
})
print("Original Sales Data:")
print(sales_data)

print("\n")

# Group by 'Product' and calculate total sales
product_sales = sales_data.groupby('Product')['Sales'].sum()
print("Total Sales by Product:")
print(product_sales)

print("\n")

# Group by 'Region' and calculate average sales
region_avg_sales = sales_data.groupby('Region')['Sales'].mean()
print("Average Sales by Region:")
print(region_avg_sales)

print("\n")

# Group by multiple columns and apply multiple aggregation functions
multi_agg = sales_data.groupby(['Product', 'Region']).agg(
    Total_Sales=('Sales', 'sum'),
    Average_Sales=('Sales', 'mean'),
    Count_Orders=('Sales', 'count')
)
print("Aggregated Sales by Product and Region:")
print(multi_agg)
```

### Merging and Joining DataFrames
Combining DataFrames is a common operation, similar to SQL joins. `pd.merge()` is the primary function for this.

```python
df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'David']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 5, 6],
    'Score': [85, 92, 78, 90]
})

print("DataFrame 1:")
print(df1)
print("\nDataFrame 2:")
print(df2)

# Inner merge (default): only common IDs
merged_inner = pd.merge(df1, df2, on='ID', how='inner')
print("\nInner Merge (common IDs only):")
print(merged_inner)

# Left merge: all from df1, matching from df2
merged_left = pd.merge(df1, df2, on='ID', how='left')
print("\nLeft Merge (all from df1, matching from df2):")
print(merged_left)

# Outer merge: all IDs from both DataFrames
merged_outer = pd.merge(df1, df2, on='ID', how='outer')
print("\nOuter Merge (all IDs from both):")
print(merged_outer)
```

## 6. Mini-Project: Analyzing a Simple Sales Dataset

Let's put your new Pandas skills to the test with a small sales dataset. We'll perform some basic cleaning, manipulation, and analysis.

### Project Goal:
Analyze monthly sales data to find:
1.  Total sales per product.
2.  Average sales per product.
3.  Highest and lowest selling product.
4.  Sales trend over months.

First, let's create a dummy sales dataset.

```python
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Product': ['A', 'A', 'A', 'A', 'A', 'A', 'B', 'B', 'B', 'B', 'B', 'B'],
    'Units_Sold': [100, 120, 150, 110, 130, 160, 80, 90, 110, 100, 120, 130],
    'Price_per_Unit': [10, 10, 10, 10, 10, 10, 15, 15, 15, 15, 15, 15]
}
sales_df = pd.DataFrame(data)

# Calculate 'Total_Sales' column
sales_df['Total_Sales'] = sales_df['Units_Sold'] * sales_df['Price_per_Unit']

print("Sales Data:")
print(sales_df)
print("\n")
sales_df.info()
```

### Task 1: Total Sales per Product
Calculate the sum of `Total_Sales` for each `Product`.

```python
total_sales_by_product = sales_df.groupby('Product')['Total_Sales'].sum()
print("Total Sales by Product:")
print(total_sales_by_product)
```

### Task 2: Average Sales per Product
Calculate the average of `Total_Sales` for each `Product`.

```python
average_sales_by_product = sales_df.groupby('Product')['Total_Sales'].mean()
print("\nAverage Sales by Product:")
print(average_sales_by_product)
```

### Task 3: Highest and Lowest Selling Product
Identify which product has the highest and lowest total sales.

```python
highest_selling_product = total_sales_by_product.idxmax()
lowest_selling_product = total_sales_by_product.idxmin()

print(f"\nHighest Selling Product: {highest_selling_product} (Total Sales: ${total_sales_by_product.max():,.2f})")
print(f"Lowest Selling Product: {lowest_selling_product} (Total Sales: ${total_sales_by_product.min():,.2f})")
```

### Task 4: Monthly Sales Trend

To see the trend, we'll need to make sure the 'Month' column is in a proper order. For simplicity, we'll assume a chronological order for the given months. We'll group by month and sum the total sales, then visualize it.

```python
import matplotlib.pyplot as plt

# Define a specific order for months
month_order = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales_df['Month'] = pd.Categorical(sales_df['Month'], categories=month_order, ordered=True)

# Group by month and sum total sales
monthly_sales = sales_df.groupby('Month')['Total_Sales'].sum().reset_index()

print("\nMonthly Sales Trend:")
print(monthly_sales)

# Plotting the trend
plt.figure(figsize=(10, 6))
plt.plot(monthly_sales['Month'], monthly_sales['Total_Sales'], marker='o')
plt.title('Total Sales Trend Over Months')
plt.xlabel('Month')
plt.ylabel('Total Sales ($)')
plt.grid(True)
plt.show()
```
