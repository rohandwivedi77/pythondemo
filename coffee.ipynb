```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Coffee':['Espresso','Latte','Cappuccino','Mocha','Cold Coffee'],
    'Cups Sold' : [120,95,150,80,110],
    'Price (₹)' : [120,180,160,200,150]
}
df = pd.DataFrame(data)
print(df)
```

```python
# Revenue
df['Revenue'] = df['Cups Sold'] * df['Price (₹)']
df
```

```python
most_sold_coffee = df.loc[df['Cups Sold'].idxmax(), 'Coffee']
most_expensive_coffee = df.loc[df['Price (₹)'].idxmax(), 'Coffee']
average_coffee_price = df['Price (₹)'].mean()
total_revenue = df['Revenue'].sum()
```

```python
print("Most coffee sold :",most_sold_coffee)
print("Most expensive coffee :",most_expensive_coffee)
print("Average coffee price :",average_coffee_price)
print("Total Revnue :",total_revenue)
```
