# Python Collections - Complete Beginner Notebook

## List - Create

```python
fruits=["Apple","Banana","Mango"]
print(fruits)
```

## List - Access

```python
fruits=["Apple","Banana","Mango"]
print(fruits[0])
print(fruits[1])
print(fruits[2])
```

## List - Negative Index

```python
fruits=["Apple","Banana","Mango"]
print(fruits[-1])
print(fruits[-2])
```

## List - Update

```python
fruits=["Apple","Banana","Mango"]
fruits[1]="Orange"
print(fruits)
```

## List - Append

```python
fruits=["Apple","Banana"]
fruits.append("Mango")
print(fruits)
```

## List - Insert

```python
fruits=["Apple","Banana"]
fruits.insert(1,"Orange")
print(fruits)
```

## List - Remove

```python
fruits=["Apple","Banana","Mango"]
fruits.remove("Banana")
print(fruits)
```

## List - Pop

```python
fruits=["Apple","Banana","Mango"]
fruits.pop()
print(fruits)
```

## List - Functions

```python
marks=[80,95,70,88]
print(len(marks))
print(max(marks))
print(min(marks))
print(sum(marks))
```

## List - Sort

```python
marks=[80,95,70,88]
marks.sort()
print(marks)
```

## List - Reverse

```python
marks=[80,95,70,88]
marks.reverse()
print(marks)
```

## List - Slicing

```python
numbers=[10,20,30,40,50]
print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
```

## List - Membership

```python
fruits=["Apple","Banana","Mango"]
print("Apple" in fruits)
print("Orange" not in fruits)
```

## Tuple - Create

```python
numbers=(10,20,30,40)
print(numbers)
```

## Tuple - Access

```python
numbers=(10,20,30,40)
print(numbers[0])
print(numbers[-1])
```

## Tuple - Count

```python
numbers=(10,20,20,30)
print(numbers.count(20))
```

## Tuple - Index

```python
numbers=(10,20,30,40)
print(numbers.index(30))
```

## Tuple - Length

```python
numbers=(10,20,30,40)
print(len(numbers))
```

## Tuple - Packing

```python
student=("Rahul",20,"Python")
print(student)
```

## Tuple - Unpacking

```python
student=("Rahul",20,"Python")
name,age,course=student
print(name)
print(age)
print(course)
```

```python
numbers={10,20,30,40}
print(numbers)
```

## Set - Union

## Set - Intersection

## Set - Difference

## Dictionary - Create

```python
student={"Name":"Rahul","Age":20,"Course":"Python"}
print(student)
```

## Dictionary - Access

```python
student={"Name":"Rahul","Age":20}
print(student["Name"])
print(student["Age"])
```

## Dictionary - get()

```python
student={"Name":"Rahul","Age":20}
print(student.get("Name"))
```

## Dictionary - Add

```python
student={"Name":"Rahul","Age":20}
student["City"]="Delhi"
print(student)
```

## Dictionary - Update

```python
student={"Name":"Rahul","Age":20}
student["Age"]=21
print(student)
```

## Dictionary - Delete

```python
student={"Name":"Rahul","Age":20,"City":"Delhi"}
del student["City"]
print(student)
```

## Dictionary - keys()

```python
student={"Name":"Rahul","Age":20}
print(student.keys())
```

## Dictionary - values()

```python
student={"Name":"Rahul","Age":20}
print(student.values())
```

## Dictionary - items()

```python
student={"Name":"Rahul","Age":20}
print(student.items())
```

## Dictionary - Membership

```python
student={"Name":"Rahul","Age":20}
print("Name" in student)
print("City" in student)
```
